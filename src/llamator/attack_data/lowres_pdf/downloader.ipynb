```python
from datasets import load_dataset
from pathlib import Path

rescale = 0.25

images_dir = Path().resolve() / "images"
images_dir.mkdir(parents=True, exist_ok=True)

dataset = load_dataset("Svyatoslav12345/lowres_docs", split="train")

for i, item in enumerate(dataset):
    # Пример при сохранении из HF
    filename = f"hfpage_{i+1}__rescale_{rescale}".replace('.', '_') + ".png"
    item["image"].save(images_dir / filename)
```
